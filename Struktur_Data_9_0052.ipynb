```python
class SimpleHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        self.table[index] = value

    def search(self, key):
        index = self.hash_function(key)
        return self.table[index]

# Contoh penggunaan
ht = SimpleHashTable(10)
ht.insert("apel", 100)
ht.insert("pisang", 200)

print("Nilai untuk 'apel':", ht.search("apel"))
print("Nilai untuk 'pisang':", ht.search("pisang"))
```

```python
class LinearProbingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        original_index = index

        while self.table[index] is not None:
            if self.table[index][0] == key:
                break
            index = (index + 1) % self.size
            if index == original_index:
                raise Exception("Hash table penuh")

        self.table[index] = (key, value)

    def search(self, key):
        index = self.hash_function(key)
        original_index = index

        while self.table[index] is not None:
            if self.table[index][0] == key:
                return self.table[index][1]
            index = (index + 1) % self.size
            if index == original_index:
                break
        return None

# Contoh penggunaan
ht = LinearProbingHashTable(5)
ht.insert("apel", 100)
ht.insert("pisang", 200)
ht.insert("melon", 300)

print("Nilai untuk 'apel':", ht.search("apel"))
print("Nilai untuk 'melon':", ht.search("melon"))
```

```python
class ChainingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [[] for _ in range(size)]

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        for i, (k, v) in enumerate(self.table[index]):
            if k == key:
                self.table[index][i] = (key, value)
                return
        self.table[index].append((key, value))

    def search(self, key):
        index = self.hash_function(key)
        for k, v in self.table[index]:
            if k == key:
                return v
        return None

# Contoh penggunaan
ht = ChainingHashTable(5)
ht.insert("apel", 100)
ht.insert("pisang", 200)
ht.insert("melon", 300)

print("Nilai untuk 'apel':", ht.search("apel"))
print("Nilai untuk 'melon':", ht.search("melon"))
```

```python
class LinearProbingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        original_index = index

        while self.table[index] is not None:
            if self.table[index][0] == key:
                break
            index = (index + 1) % self.size
            if index == original_index:
                raise Exception("Hash table penuh")

        self.table[index] = (key, value)

    def search(self, key):
        index = self.hash_function(key)
        original_index = index

        while self.table[index] is not None:
            if self.table[index][0] == key:
                return self.table[index][1]
            index = (index + 1) % self.size
            if index == original_index:
                break
        return None

# a. Buat objek dengan ukuran tabel 5
ht = LinearProbingHashTable(5)

# b. Masukkan 5 pasangan key-value
data = {"A": 10, "B": 20, "C": 30, "D": 40, "E": 50}
for key, value in data.items():
    ht.insert(key, value)

# c. Cetak isi array self.table
print("=== Isi self.table setelah semua data dimasukkan ===")
for i, slot in enumerate(ht.table):
    print(f"  Index {i}: {slot}")

# d. Penjelasan collision
print()
print("=== Penjelasan: Jika dua key memiliki hasil hash yang sama ===")
print(f"  hash('A') % 5 = {hash('A') % 5}")
print(f"  hash('B') % 5 = {hash('B') % 5}")
print(f"  hash('C') % 5 = {hash('C') % 5}")
print(f"  hash('D') % 5 = {hash('D') % 5}")
print(f"  hash('E') % 5 = {hash('E') % 5}")
print()
print("  Jika dua key menghasilkan indeks yang sama (collision),")
print("  Linear Probing akan mencari slot kosong berikutnya secara")
print("  berurutan: (index + 1) % size, (index + 2) % size, dst.")
print("  Data tetap tersimpan tanpa saling menimpa, namun jika")
print("  terlalu banyak collision dapat terjadi 'clustering' yang")
print("  menurunkan efisiensi pencarian.")
```

```python
class ChainingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [[] for _ in range(size)]

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        for i, (k, v) in enumerate(self.table[index]):
            if k == key:
                self.table[index][i] = (key, value)
                return
        self.table[index].append((key, value))

    def search(self, key):
        index = self.hash_function(key)
        for k, v in self.table[index]:
            if k == key:
                return v
        return None

# a. Gunakan ukuran tabel 3
ht = ChainingHashTable(3)

# b. Masukkan key-value berikut
ht.insert("apel", 100)
ht.insert("melon", 200)
ht.insert("lemon", 300)
ht.insert("pisang", 400)

# c. Cetak isi tabel dan semua data di setiap slot
print("=== Isi self.table (ChainingHashTable ukuran 3) ===")
for i, slot in enumerate(ht.table):
    print(f"  Slot {i}: {slot}")

print()
print("=== Hash masing-masing key ===")
for key in ["apel", "melon", "lemon", "pisang"]:
    print(f"  hash('{key}') % 3 = {hash(key) % 3}")

# d. Penjelasan collision pada chaining
print()
print("=== Penjelasan: Bagaimana collision ditangani dalam Chaining ===")
print("  Berbeda dengan Linear Probing, Chaining tidak mencari slot lain.")
print("  Setiap slot menyimpan sebuah list (rantai / chain).")
print("  Ketika dua key menghasilkan indeks yang sama (collision),")
print("  key baru cukup di-append ke list di slot tersebut.")
print("  Saat pencarian, program menelusuri list di slot yang dituju")
print("  satu per satu sampai key yang cocok ditemukan.")
print("  Keunggulan: tidak ada clustering & tabel tidak bisa 'penuh'")
print("  selama memori masih tersedia.")
```

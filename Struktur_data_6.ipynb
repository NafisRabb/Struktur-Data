```python
def bubble_sort(arr):
  n = len(arr)
  for i in range(n - 1):
    for j in range(n - i - 1):
      if arr[j] > arr[j + 1]:
        arr[j], arr[j + 1] = arr[j + 1], arr[j]
  return arr

data = [64,34,25,12,22,11,90]
sorted_data = bubble_sort(data.copy())
print("Data sebelum sorting ", data)
print("Data setelah sorting ", sorted_data)





```

```python
def selection_sort(arr):
  n = len(arr)
  for i in range(n-1):
    min_index = i
    for j in range(i+1,n):
      if arr[j] < arr[min_index]:
        min_index = j
    arr[i], arr[min_index] = arr[min_index], arr[i]
  return arr

data = [54,25,12,22,11]
sorted_data = selection_sort(data.copy())
print("sebelum sorting", data)
print("setelah sorting", sorted_data)







```

```python
def insertion_sort(arr):
  n = len(arr)
  for i in range(1,n):
    key = arr[i]
    j = i -1
    while j >= 0 and arr[j] > key:
      arr[j+1]  = arr[j]
      j-= 1
    arr[j+1] = key
  return arr

data = [12,11,14,15,13]
sorted_data = insertion_sort(data.copy())
print("data sebelum sorting", data)
print("data sesudah sorting", sorted_data)

```

```python
def merge_sort(arr):
  if len(arr) > 1:
    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    merge_sort(left_half)
    merge_sort(right_half)

    i = j = k = 0

    while i <len(left_half) and j < len(right_half):
      if left_half[i] < right_half[j]:
          arr[k] = left_half[i]
          i += 1
      else:
          arr[k] = right_half[j]
          j += 1
          k += 1
    while i < len(left_half):
      arr[k] = left_half[i]
      i += 1
      k += 1

    while j < len(right_half):
      arr[k] = right_half[j]
      j += 1
      k += 1
  return arr

data = [11,15,17,12,13,2]
sorted_data = merge_sort(data.copy())
print("data sebelum sorting", data)
print("data sesudah sorting", sorted_data)
```

```python
def quick_sort(arr):
    if len(arr) <= 1:
      return arr
    pivot = arr[len(arr) //2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

data = [12,11,16,21,34,2]
sorted_data = quick_sort(data.copy())
print("data sebelum sorting", data)
print("data sesudah sorting", sorted_data)
```

```python
# Data Mahasiswa Rombel 03
mahasiswa = [
    {"nama": "Budi", "nilai": 85},
    {"nama": "Ani", "nilai": 92},
    {"nama": "Caca", "nilai": 78},
    {"nama": "Dedi", "nilai": 95},
    {"nama": "Euis", "nilai": 88}
]

def selection_sort_descending(data):
    n = len(data)
    for i in range(n):
        # Anggap indeks saat ini sebagai nilai tertinggi
        max_idx = i

        # Cari nilai yang lebih besar di sisa list
        for j in range(i + 1, n):
            if data[j]["nilai"] > data[max_idx]["nilai"]:
                max_idx = j

        # Tukar posisi jika ditemukan nilai yang lebih tinggi
        data[i], data[max_idx] = data[max_idx], data[i]

def tampilkan_peringkat(data):
    print(f"{'Peringkat':<10} | {'Nama':<10} | {'Nilai':<5}")
    print("-" * 30)
    for i, mhs in enumerate(data, 1):
        print(f"{i:<10} | {mhs['nama']:<10} | {mhs['nilai']:<5}")

# Eksekusi
print("Mengurutkan data...\n")
selection_sort_descending(mahasiswa)
tampilkan_peringkat(mahasiswa)
```

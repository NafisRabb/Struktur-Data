```python
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

def print_graph(graf):
    for node in graf:
        print(f"{node} -> {graf[node]}")

print_graph(graph)
```

```python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])

    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            print(vertex, end=" ")
            visited.add(vertex)
            queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

# Contoh pemanggilan
bfs(graph, 'A')
```

```python
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    if start not in visited:
        print(start, end=" ")
        visited.add(start)
        for neighbor in graph[start]:
            dfs(graph, neighbor, visited)

# Contoh pemanggilan
dfs(graph, 'A')
```

```python
# Lengkapi fungsi untuk menghitung jumlah tetangga dari tiap simpul dalam graf
def count_neighbors(graph):
    result = {}
    for node in graph:
        # Hitung jumlah tetangga
        result[node] = len(graph[node])
    return result

# ── Contoh Graf ──────────────────────────────────────────
graph = {
    'A': ['B', 'C', 'D'],
    'B': ['A', 'C'],
    'C': ['A', 'B'],
    'D': ['A']
}

# ── Jalankan fungsi ──────────────────────────────────────
hasil = count_neighbors(graph)

# ── Tampilkan hasil ──────────────────────────────────────
print("Jumlah tetangga tiap simpul:")
for node, count in hasil.items():
    print(f"  Simpul {node} : {count} tetangga")
```

```python
from collections import deque

# ============================================================
# TUGAS 1: Graf tak berarah dengan minimal 5 simpul dan 7 sisi
# ============================================================

def buat_graf():
    """Membuat graf tak berarah menggunakan adjacency list."""
    graf = {
        'A': ['B', 'C', 'D'],
        'B': ['A', 'C', 'E'],
        'C': ['A', 'B', 'D', 'F'],
        'D': ['A', 'C', 'E'],
        'E': ['B', 'D', 'F'],
        'F': ['C', 'E'],
    }
    return graf

def print_graph(graf):
    """Menampilkan graf dalam format adjacency list."""
    print("=" * 40)
    print("       GRAF TAK BERARAH")
    print("=" * 40)
    jumlah_sisi = 0
    sisi_dihitung = set()
    for simpul, tetangga in graf.items():
        print(f"  {simpul} --> {', '.join(tetangga)}")
        for t in tetangga:
            sisi = tuple(sorted([simpul, t]))
            sisi_dihitung.add(sisi)
    jumlah_sisi = len(sisi_dihitung)
    print("-" * 40)
    print(f"  Jumlah simpul : {len(graf)}")
    print(f"  Jumlah sisi   : {jumlah_sisi}")
    print("=" * 40)


# ============================================================
# TUGAS 2: Implementasi BFS dan DFS
# ============================================================

def bfs(graf, simpul_awal):
    """Breadth-First Search — mencetak urutan kunjungan."""
    dikunjungi = set()
    antrian = deque([simpul_awal])
    dikunjungi.add(simpul_awal)

    print(f"\nBFS dari simpul '{simpul_awal}':")
    urutan = []
    while antrian:
        simpul = antrian.popleft()
        urutan.append(simpul)
        for tetangga in graf[simpul]:
            if tetangga not in dikunjungi:
                dikunjungi.add(tetangga)
                antrian.append(tetangga)

    print("  Urutan kunjungan:", " -> ".join(urutan))

def dfs(graf, simpul_awal, dikunjungi=None, urutan=None, _cetak=True):
    """Depth-First Search — mencetak urutan kunjungan."""
    if dikunjungi is None:
        dikunjungi = set()
        urutan = []
        if _cetak:
            print(f"\nDFS dari simpul '{simpul_awal}':")

    dikunjungi.add(simpul_awal)
    urutan.append(simpul_awal)

    for tetangga in graf[simpul_awal]:
        if tetangga not in dikunjungi:
            dfs(graf, tetangga, dikunjungi, urutan, _cetak=False)

    if _cetak:
        print("  Urutan kunjungan:", " -> ".join(urutan))

    return urutan, dikunjungi


# ============================================================
# TUGAS 3: BFS yang mengembalikan list urutan kunjungan
# ============================================================

def bfs_urutan(graf, simpul_awal):
    """BFS yang mengembalikan LIST urutan kunjungan (bukan hanya mencetak)."""
    dikunjungi = set()
    antrian = deque([simpul_awal])
    dikunjungi.add(simpul_awal)
    urutan = []

    while antrian:
        simpul = antrian.popleft()
        urutan.append(simpul)
        for tetangga in graf[simpul]:
            if tetangga not in dikunjungi:
                dikunjungi.add(tetangga)
                antrian.append(tetangga)

    return urutan  # mengembalikan list


# ============================================================
# TUGAS 4: find_path menggunakan DFS
# ============================================================

def find_path(graf, start, end, dikunjungi=None, jalur=None):
    """
    Mencari SATU jalur dari 'start' ke 'end' menggunakan DFS.
    Mengembalikan list jalur jika ditemukan, None jika tidak ada.
    """
    if dikunjungi is None:
        dikunjungi = set()
        jalur = []

    dikunjungi.add(start)
    jalur = jalur + [start]

    if start == end:
        return jalur  # jalur ditemukan

    for tetangga in graf[start]:
        if tetangga not in dikunjungi:
            hasil = find_path(graf, tetangga, end, dikunjungi, jalur)
            if hasil is not None:
                return hasil

    return None  # tidak ada jalur


# ============================================================
# TUGAS 5: is_connected menggunakan DFS + logika
# ============================================================

def is_connected(graf):
    """
    Mengecek apakah semua simpul dalam graf saling terhubung.
    Menggunakan DFS dari simpul pertama, lalu cek apakah semua simpul terkunjungi.
    """
    if not graf:
        return True  # graf kosong dianggap terhubung

    simpul_awal = next(iter(graf))  # ambil simpul pertama
    _, dikunjungi = dfs(graf, simpul_awal, _cetak=False)

    # Cek apakah semua simpul terkunjungi
    semua_simpul = set(graf.keys())
    return dikunjungi == semua_simpul


# ============================================================
# MAIN — Demonstrasi semua tugas
# ============================================================

if __name__ == "__main__":

    # Tugas 1: Buat dan tampilkan graf
    print("\n>>> TUGAS 1: Buat Graf & print_graph")
    graf = buat_graf()
    print_graph(graf)

    # Tugas 2: BFS dan DFS
    print("\n>>> TUGAS 2: BFS dan DFS")
    bfs(graf, 'A')
    dfs(graf, 'A')

    # Tugas 3: BFS yang mengembalikan list
    print("\n>>> TUGAS 3: BFS mengembalikan list urutan kunjungan")
    hasil_bfs = bfs_urutan(graf, 'A')
    print(f"  bfs_urutan('A') = {hasil_bfs}")
    print(f"  Tipe data       : {type(hasil_bfs)}")

    # Tugas 4: find_path dengan DFS
    print("\n>>> TUGAS 4: find_path(graf, start, end) dengan DFS")
    pasangan = [('A', 'F'), ('B', 'F'), ('E', 'A'), ('F', 'B')]
    for s, e in pasangan:
        jalur = find_path(graf, s, e)
        if jalur:
            print(f"  {s} -> {e} : {' -> '.join(jalur)}")
        else:
            print(f"  {s} -> {e} : Tidak ada jalur")

    # Tugas 5: is_connected
    print("\n>>> TUGAS 5: is_connected(graph)")
    terhubung = is_connected(graf)
    print(f"  Graf terhubung? : {terhubung}")

    # Uji dengan graf terputus
    graf_terputus = {
        'A': ['B'],
        'B': ['A'],
        'C': ['D'],  # komponen terpisah
        'D': ['C'],
    }
    print(f"  Graf terputus terhubung? : {is_connected(graf_terputus)}")
```

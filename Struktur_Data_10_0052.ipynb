```python
import heapq
import networkx as nx
import matplotlib.pyplot as plt

def dijkstra_with_paths(graph, start):
    distances = {node: float('inf') for node in graph}
    distances[start] = 0
    previous = {node: None for node in graph}
    queue = [(0, start)]

    while queue:
        current_distance, current_node = heapq.heappop(queue)

        for neighbor, weight in graph[current_node].items():
            distance = current_distance + weight
            if distance < distances[neighbor]:
                distances[neighbor] = distance
                previous[neighbor] = current_node
                heapq.heappush(queue, (distance, neighbor))

    return distances, previous
```

```python
def get_path(previous, target):
    path = []
    while target is not None:
        path.insert(0, target)
        target = previous[target]
    return path
```

```python
def visualize_graph(graph, path=None, title="Visualisasi Graf dan Jalur Terpendek", layout="spring"):
    G = nx.DiGraph()

    for node in graph:
        for neighbor, weight in graph[node].items():
            G.add_edge(node, neighbor, weight=weight)

    if layout == "spring":
        pos = nx.spring_layout(G, seed=42)
    elif layout == "circular":
        pos = nx.circular_layout(G)
    elif layout == "kamada_kawai":
        pos = nx.kamada_kawai_layout(G)
    else:
        pos = nx.spring_layout(G, seed=42)

    edge_labels = nx.get_edge_attributes(G, 'weight')

    plt.figure(figsize=(10, 7))
    nx.draw(G, pos, with_labels=True, node_color='lightblue', node_size=2000,
            font_weight='bold', arrows=True)
    nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

    # Garis tebal untuk jalur terpendek
    if path and len(path) > 1:
        path_edges = list(zip(path, path[1:]))
        nx.draw_networkx_edges(G, pos, edgelist=path_edges, edge_color='red', width=3)

    plt.title(title)
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
# Definisi graf
graph = {
    'A': {'B': 4, 'C': 2},
    'B': {'C': 1, 'D': 5},
    'C': {'D': 8, 'E': 10},
    'D': {'E': 2, 'Z': 6},
    'E': {'Z': 3},
    'Z': {}
}

# Jalankan Dijkstra
start_node = 'A'
end_node = 'Z'
distances, previous = dijkstra_with_paths(graph, start_node)

# Rekonstruksi jalur
shortest_path = get_path(previous, end_node)
print(f"Jarak dari {start_node} ke {end_node}: {distances[end_node]}")
print(f"Jalur: {' → '.join(shortest_path)}")

# Visualisasi
visualize_graph(graph, path=shortest_path)
```

```python
graph_siklus = {
    'A': {'B': 4, 'C': 2},
    'B': {'C': 1, 'D': 5},
    'C': {'D': 8, 'E': 10, 'A': 3},   # C → A membentuk siklus
    'D': {'E': 2, 'Z': 6, 'B': 1},    # D → B membentuk siklus
    'E': {'Z': 3},
    'Z': {}
}

distances_s, previous_s = dijkstra_with_paths(graph_siklus, start_node)
shortest_path_s = get_path(previous_s, end_node)

print(f"Jarak dari {start_node} ke {end_node}: {distances_s[end_node]}")
print(f"Jalur: {' → '.join(shortest_path_s)}")

visualize_graph(
    graph_siklus,
    path=shortest_path_s,
    title="Tugas 1 – Graf dengan Siklus (C→A dan D→B)"
)
```

```python
graph_baru = {
    'A': {'B': 4, 'C': 2, 'G': 3},   # A terhubung ke simpul baru G
    'B': {'C': 1, 'D': 5},
    'C': {'D': 8, 'E': 10},
    'D': {'E': 2, 'Z': 6},
    'E': {'Z': 3},
    'G': {'F': 1},                    # Simpul baru G
    'F': {'Z': 1},                    # Simpul baru F
    'Z': {}
}

distances_b, previous_b = dijkstra_with_paths(graph_baru, start_node)
shortest_path_b = get_path(previous_b, end_node)

print(f"Jarak dari {start_node} ke {end_node}: {distances_b[end_node]}")
print(f"Jalur: {' → '.join(shortest_path_b)}")
print("Rute baru A → G → F → Z lebih pendek dari rute semula!")

visualize_graph(
    graph_baru,
    path=shortest_path_b,
    title="Tugas 2 – Graf dengan Simpul Baru F & G"
)
```

```python
# Layout: circular
visualize_graph(
    graph,
    path=shortest_path,
    title="Tugas 3a – Layout: circular_layout",
    layout="circular"
)
```

```python
# Layout: kamada_kawai
visualize_graph(
    graph,
    path=shortest_path,
    title="Tugas 3b – Layout: kamada_kawai_layout",
    layout="kamada_kawai"
)
```
